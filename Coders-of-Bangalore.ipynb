```python
with open ('Data/finaldata.txt', encoding='utf-8') as f:
    data = f.read()
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)>3]
```

```python
def parse_chunk(chunk):
   # try:
        chunk = chunk.strip()
        sep_chunk = chunk.split("\n")
        
        username = sep_chunk[0]
        
        no_of_post = int(sep_chunk[1].split(" post")[0].replace(",", ""))
        
        no_of_followers = float(sep_chunk[2].split(" followers")[0].replace(",", "").replace("K", "").replace("M", ""))
        if("K" in sep_chunk[2]):
            no_of_followers = int(no_of_followers * 1000)
        elif ("M" in sep_chunk[2]):
            no_of_followers = int(no_of_followers * 1000000)
        else:
            no_of_followers = int(no_of_followers)
        
        no_of_following = float(sep_chunk[3].split(" following")[0].replace(",", "").replace("K", "").replace("M", ""))
        if("K" in sep_chunk[3]):
           no_of_following = int(no_of_following * 1000)
        elif ("M" in sep_chunk[3]):
            no_of_following = int(no_of_following * 1000000)
        else:
            no_of_following = int(no_of_following)
        
        name = sep_chunk[4]
        
        if (len(sep_chunk)>5):
            type_of_categories = sep_chunk[5]
            bio = "\n".join(sep_chunk[6:])
        else:
            type_of_categories = "Unknown"
            bio = ""
        # print(username, no_of_post, no_of_followers, no_of_following, name, 
        #       bio, sep='\n')
        return{"username":username, "no_of_post":no_of_post, "no_of_followers":no_of_followers, 
               "no_of_following":no_of_following, "name":name, "type_of_categories":type_of_categories, "bio":bio}
    # except Exception as e:
    #     print(e, chunk)
# print(parse_chunk(chunks[0]))
    
```

```python
all_chunks = []
for chunk in chunks:
    # print(f"The chunk is:{chunk} \n")
    prased_chunk = parse_chunk(chunk)
    all_chunks.append(prased_chunk)
```

```python
import json

s = json.dumps(all_chunks, indent = 4)
with open("Data/data.json", "w") as f:
    f.write(s)
```

# Who has the mask followers

```python
max_post = 0 
for chunk in all_chunks:
    if (max_post<chunk['no_of_post']):
        max_post = chunk['no_of_post']
        chunk_with_max_post = chunk

print(chunk_with_max_post['username'])
```

# Who has Maximum Followers

```python
max_follower = 0

for chunk in all_chunks:
    if (max_follower<chunk['no_of_followers']):
        max_follower = chunk['no_of_followers']
        chunk_with_max_follower = chunk
print(chunk_with_max_follower['username'])
    
```

# Who follows maximum people 

```python
max_following = 0

for chunk in all_chunks:
    if max_following < chunk['no_of_following']:
        max_following = chunk['no_of_following']
        chunk_with_max_following = chunk

print(chunk_with_max_following['username'])
```

# How many categories 

```python
categories = set()

for chunk in all_chunks:
    categories.add(chunk['type_of_categories'])

print(f"There are {len(categories)} categories", sep='\n')
```
